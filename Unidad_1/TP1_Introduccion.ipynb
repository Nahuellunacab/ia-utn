## TP 1 - Introducción a la Inteligencia Artificial

---

### Defina con sus propias palabras: inteligencia, inteligencia artificial

La inteligencia puede entenderse como la capacidad de una persona (o sistema) para aprender, razonar, adaptarse a nuevas situaciones y resolver problemas a partir de la información disponible. No se trata solo de saber cosas, sino de poder usar ese conocimiento de forma efectiva en distintos contextos.

Por otro lado, la inteligencia artificial (IA) es un área de la informática que busca desarrollar sistemas capaces de realizar tareas que normalmente requieren inteligencia humana. Esto incluye actividades como reconocer patrones, tomar decisiones, aprender de la experiencia o incluso comprender lenguaje natural.

---

### En realidad los computadores no son inteligentes, hacen solamente lo que dicen los programadores. ¿Es cierta esta aseveración, e implica a la primera?

La afirmación de que los computadores no son inteligentes porque solo hacen lo que les indican los programadores es parcialmente cierta, pero incompleta. Es verdad que todo sistema informático parte de instrucciones humanas, sin embargo, muchos sistemas actuales de IA pueden aprender a partir de datos y mejorar su desempeño sin que se les indique explícitamente cada paso.

Por ejemplo, en el aprendizaje automático, los algoritmos no siguen instrucciones rígidas, sino que ajustan su comportamiento en función de la experiencia. En ese sentido, aunque no poseen inteligencia en el mismo sentido que un humano, sí pueden exhibir comportamientos que parecen inteligentes.

Esto no contradice la definición de inteligencia artificial, sino que la complementa: la IA no busca replicar exactamente la inteligencia humana, sino simular ciertos aspectos de ella mediante sistemas computacionales.

### Consulte en la literatura existente sobre la IA si alguna de las siguientes tareas se puede efectuar con computadoras: a) Jugar una partida de tenis de mesa (ping-pong) decentemente. b) Conducir un coche en el centro del Cairo. c) Comprar comestibles para una semana en el mercado. d) Comprar comestibles para una semana en la web. e) Descubrir y demostrar nuevos teoremas matemáticos. f) Escribir intencionadamente una historia divertida. g) Ofrecer asesoría legal competente en un área determinada. h) Traducir inglés hablado al sueco hablado en tiempo real. i) Realizar una operación de cirugía compleja.

Actualmente, muchas de las tareas mencionadas pueden ser realizadas por computadoras, aunque con distintos niveles de éxito.

Jugar al ping-pong es posible mediante robots especializados, aunque todavía con ciertas limitaciones en entornos reales. Conducir un coche también es una tarea que ya realizan sistemas autónomos, pero en contextos complejos como el centro de El Cairo sigue siendo un desafío importante.

Comprar comestibles en un mercado físico es difícil porque requiere percepción del entorno, interacción y toma de decisiones en tiempo real. En cambio, comprar en la web es totalmente viable y ampliamente automatizado hoy en día.

En cuanto a descubrir y demostrar teoremas matemáticos, existen sistemas capaces de hacerlo, incluso generando nuevos resultados. Escribir una historia divertida también es posible con modelos de lenguaje, aunque el humor sigue siendo un aspecto complejo.

La asesoría legal puede ser parcialmente automatizada en áreas específicas, pero no reemplaza completamente a un profesional. La traducción en tiempo real es una tecnología bastante avanzada y muy utilizada actualmente. Por último, en cirugía compleja existen robots que asisten en operaciones, pero siempre bajo supervisión humana.

En general, la IA ha avanzado mucho, pero el nivel de desempeño depende fuertemente del contexto y la complejidad del entorno.

### Explique sintéticamente las cuatro categorías que se vienen siguiendo respecto de la definición de inteligencia artificial.

Existen distintas formas de definir la inteligencia artificial, que se agrupan en cuatro enfoques principales.

Por un lado, están los sistemas que piensan como humanos, que buscan imitar los procesos mentales, como el razonamiento o la toma de decisiones. Por otro lado, los sistemas que actúan como humanos intentan comportarse de manera indistinguible de una persona, como en el caso del Test de Turing.

También existen los sistemas que piensan racionalmente, que se basan en la lógica formal para llegar a conclusiones correctas. Finalmente, los sistemas que actúan racionalmente se enfocan en tomar la mejor decisión posible en función de los objetivos y la información disponible.

Cada uno de estos enfoques ofrece una perspectiva diferente sobre qué significa que un sistema sea inteligente.
