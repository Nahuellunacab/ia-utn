## TP 2 - Agentes Inteligentes

---

### Defina con sus propias palabras los siguientes términos: agente, función de agente, programa de agente, racionalidad, autonomía, agente reactivo, agente basado en modelo, agente basado en objetivos, agente basado en utilidad, agente que aprende. 

Un agente puede entenderse como cualquier entidad que percibe su entorno a través de sensores y actúa sobre él mediante actuadores. Es decir, recibe información, la procesa y toma decisiones.

La función de agente es la relación que define qué acción debe realizar el agente en función de lo que ha percibido hasta el momento. En cambio, el programa de agente es la implementación concreta de esa función, es decir, el código que ejecuta esas decisiones.

La racionalidad se refiere a la capacidad del agente de elegir la mejor acción posible en función de sus objetivos y la información disponible. Por otro lado, la autonomía indica qué tan independiente es el agente respecto del conocimiento previo que le haya sido dado, es decir, cuánto puede aprender por sí mismo.

Un agente reactivo actúa únicamente en función de la percepción actual, sin considerar estados pasados. En cambio, un agente basado en modelo mantiene una representación interna del entorno para tomar decisiones más informadas.

Un agente basado en objetivos toma decisiones en función de metas específicas que quiere alcanzar, mientras que un agente basado en utilidad evalúa qué tan “bueno” es cada resultado posible y elige el más conveniente.

Finalmente, un agente que aprende es aquel que puede mejorar su desempeño a partir de la experiencia, adaptándose con el tiempo.

### dentifique la descripción REAS que define el entorno de trabajo para cada uno de los siguientes agentes: a) Robot que juega al fútbol; b) Agente para comprar libros en Internet; c) Explorador autónomo de Marte; d) Asistente matemático para la demostración de teoremas. 

Para describir el entorno de trabajo de cada agente utilizamos REAS (Rendimiento, Entorno, Actuadores y Sensores).

a) Robot que juega al fútbol:  
- Rendimiento: ganar el partido o maximizar goles  
- Entorno: cancha, pelota, jugadores  
- Actuadores: motores, piernas, mecanismos de movimiento  
- Sensores: cámaras, sensores de posición, proximidad  

b) Agente para comprar libros en Internet:  
- Rendimiento: comprar el mejor libro al mejor precio  
- Entorno: sitios web, tiendas online  
- Actuadores: acciones de navegación y compra  
- Sensores: información de precios, disponibilidad, reseñas  

c) Explorador autónomo de Marte:  
- Rendimiento: recolectar información útil del entorno  
- Entorno: superficie marciana  
- Actuadores: ruedas, brazos robóticos  
- Sensores: cámaras, sensores térmicos, químicos  

d) Asistente matemático:  
- Rendimiento: demostrar teoremas correctamente  
- Entorno: problemas matemáticos  
- Actuadores: generación de pasos lógicos  
- Sensores: entrada de datos o enunciados

a) Robot que juega al fútbol:  
Parcialmente observable, estocástico, secuencial, dinámico, continuo, multiagente.

b) Agente de compra online:  
Parcialmente observable, estocástico, secuencial, dinámico, discreto, multiagente.

c) Explorador de Marte:  
Parcialmente observable, estocástico, secuencial, dinámico, continuo, agente individual.

d) Asistente matemático:  
Totalmente observable, determinista, secuencial, estático, discreto, agente individual.

### Represente a través de diagramas esquemáticos los cuatro tipos de modelos de agentes.

Los principales modelos de agentes pueden representarse de la siguiente forma:

- Agente reactivo: percibe el entorno y responde directamente sin memoria.
- Agente basado en modelo: mantiene un estado interno del mundo.
- Agente basado en objetivos: toma decisiones en función de metas a alcanzar.
- Agente basado en utilidad: evalúa qué resultado es mejor según una función de utilidad.

Estos modelos representan distintos niveles de complejidad en la toma de decisiones de un agente.
